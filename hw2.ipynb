## A 排序

```python

#include <bits/stdc++.h>
using namespace std;

int main() {
    ios::sync_with_stdio(false);
    cin.tie(nullptr);

    int k, a, b;
    cin >> k;
    int n = 1 << k;
    cin >> a >> b;

    vector<int> arr(n);
    for (int i = 0; i < n; i++) {
        cin >> arr[i];
    }

    vector<int> target(n);
    iota(target.begin(), target.end(), 0);

    if (arr == target) {
        cout << "0\n";
        return 0;
    }

    vector<pair<int, int>> ops;

    // 核心思想：通过加法循环移位，配合交换，可以实现任意相邻元素的交换
    // 从而可以实现冒泡排序

    // 对于每个位置i，确保arr[i] == i
    for (int i = 0; i < n; i++) {
        if (arr[i] == i) continue;

        // 找到i在哪
        int pos = find(arr.begin(), arr.end(), i) - arr.begin();

        // 我们需要将pos位置的元素移动到i位置
        // 通过循环移位和交换来实现

        // 计算需要移动的距离
        int dist = (pos - i + n) % n;

        // 通过加法将元素移到a,b位置，交换，再移回来
        // 简化：我们可以直接使用加法循环，配合交换，实现相邻交换

        // 这里使用示例中的模式：多次加法配合一次交换
        for (int step = 0; step < dist; step++) {
            // 加1操作
            ops.push_back({2, 1});
            for (int j = 0; j < n; j++) {
                arr[j] = (arr[j] + 1) % n;
            }
        }

        // 现在需要交换a和b
        ops.push_back({0, 0});
        int posA = find(arr.begin(), arr.end(), a) - arr.begin();
        int posB = find(arr.begin(), arr.end(), b) - arr.begin();
        swap(arr[posA], arr[posB]);

        // 逆向加法
        for (int step = 0; step < dist; step++) {
            ops.push_back({2, n-1}); // 加n-1相当于减1
            for (int j = 0; j < n; j++) {
                arr[j] = (arr[j] + n - 1) % n;
            }
        }
    }

    if (arr == target) {
        cout << ops.size() << "\n";
        for (auto [type, x] : ops) {
            if (type == 0) cout << "0\n";
            else cout << type << " " << x << "\n";
        }
    } else {
        cout << "-1\n";
    }

    return 0;
}
```

## B 长跑

```python
#include <iostream>
#include <cstdio>
#include <cstring>
#include <algorithm>
#include <queue>
#include <map>
using namespace std;

const int maxm = 2e4 + 5;

int n, l, ma, s;

struct Shop {
    int p, c; // p距离，c花费
} a[maxm];

struct Node {
    int now, money, id; // now当前距离，money剩下的钱，id当前商店位置
};

bool cmp(Shop a, Shop b) {
    if (a.p != b.p) return a.p < b.p; // 按距离起点从小到大排序
    return a.c < b.c; // 距离一样按花费排序
}

int main() {
    while (scanf("%d%d%d%d", &n, &l, &ma, &s) != EOF) {
        for (int i = 1; i <= n; i++) {
            scanf("%d%d", &a[i].p, &a[i].c);
        }

        // 如果初始体力就能直接到达终点
        if (ma >= l) {
            puts("Yes");
            continue;
        }

        // 预处理：合并同位置的商店，只保留最便宜的
        sort(a + 1, a + 1 + n, cmp);
        int cnt = 0;
        for (int i = 1; i <= n; i++) {
            if (i == 1 || a[i].p != a[i-1].p) {
                a[++cnt] = a[i];
            }
        }
        n = cnt;

        // 加入终点
        a[++n].p = l;
        a[n].c = 0;

        // BFS
        queue<Node> q;
        Node st = {0, s, 0};
        q.push(st);

        // 记忆化：记录到达某个商店时的最大剩余金钱
        vector<int> maxMoney(n + 1, -1);
        maxMoney[0] = s;

        bool ok = false;

        while (!q.empty() && !ok) {
            Node x = q.front();
            q.pop();

            // 如果当前位置的剩余金钱没有之前记录的多，跳过
            if (x.money < maxMoney[x.id]) continue;

            for (int i = x.id + 1; i <= n; i++) {
                int dist = a[i].p - x.now;

                // 大于体力值的地方不可能到
                if (dist > ma) break;

                // 如果是终点
                if (a[i].p == l) {
                    ok = true;
                    break;
                }

                // 如果钱够买补给
                if (x.money >= a[i].c) {
                    int newMoney = x.money - a[i].c;
                    // 只有剩余金钱更多时才入队
                    if (newMoney > maxMoney[i]) {
                        maxMoney[i] = newMoney;
                        Node t = {a[i].p, newMoney, i};
                        q.push(t);
                    }
                }

                // 也可以选择不补给直接路过（但通常补给更好）
                // 如果体力足够到更远的点，不补给也可能有益
                if (dist < ma) {
                    // 不补给，继续往前探索
                    // 这里不需要入队，因为会在后续循环中自然处理
                }
            }
        }

        puts(ok ? "Yes" : "No");
    }
    return 0;
}
```

## C 最长回文

```python
#include <bits/stdc++.h>
using namespace std;

const int maxn = 2e6 + 50;

char a[maxn], newa[maxn];
int p1[maxn];
char b[maxn], newb[maxn];
int p2[maxn];

// 预处理字符串A，插入分隔符
int init1() {
    int len = strlen(a);
    newa[0] = '$';
    newa[1] = '#';
    int j = 2;
    for (int i = 0; i < len; i++) {
        newa[j++] = a[i];
        newa[j++] = '#';
    }
    newa[j] = '\0';
    return j;
}

// 预处理字符串B，插入分隔符
int init2() {
    int len = strlen(b);
    newb[0] = '$';
    newb[1] = '#';
    int j = 2;
    for (int i = 0; i < len; i++) {
        newb[j++] = b[i];
        newb[j++] = '#';
    }
    newb[j] = '\0';
    return j;
}

// Manacher算法处理A
void manacher1() {
    int len = init1();
    int id = 0, mx = 0;
    for (int i = 1; i < len; i++) {
        if (i < mx)
            p1[i] = min(p1[2 * id - i], mx - i);
        else
            p1[i] = 1;

        while (newa[i - p1[i]] == newa[i + p1[i]])
            p1[i]++;

        if (mx < i + p1[i]) {
            id = i;
            mx = i + p1[i];
        }
    }
}

// Manacher算法处理B
void manacher2() {
    int len = init2();
    int id = 0, mx = 0;
    for (int i = 1; i < len; i++) {
        if (i < mx)
            p2[i] = min(p2[2 * id - i], mx - i);
        else
            p2[i] = 1;

        while (newb[i - p2[i]] == newb[i + p2[i]])
            p2[i]++;

        if (mx < i + p2[i]) {
            id = i;
            mx = i + p2[i];
        }
    }
}

int main() {
    int n;
    scanf("%d", &n);
    scanf("%s", a);
    scanf("%s", b);

    manacher1();
    manacher2();

    int ans = 1;
    // 处理后的字符串长度为 (n+1)*2
    n = (n + 1) * 2;

    // 枚举交界点（对应Manacher数组中的'#'位置）
    for (int i = 2; i <= n; i++) {
        // 初始半径取两者中的较大值
        int tmp = max(p1[i], p2[i - 2]);

        // 向两边扩展匹配
        while (newa[i - tmp] == newb[i + tmp - 2])
            tmp++;

        ans = max(ans, tmp);
    }

    printf("%d\n", ans - 1);  // 减1是因为Manacher算法的半径包含中心
    return 0;
}
```

## D 优惠券

```python
#include <iostream>
#include <set>
#include <map>
#include <cstdio>
using namespace std;

char c[555555];
int id[555555];

int main() {
    int n;
    while (~scanf("%d", &n)) {
        int ans = -1;
        set<int> pos;        // 存储可用的 '?' 的位置
        map<int, int> a;     // 记录每个券最近一次出现的位置
        char op[2];

        for (int i = 1; i <= n; i++) {
            scanf("%s", op);
            c[i] = op[0];
            if (c[i] != '?') {
                scanf("%d", &id[i]);
            }

            if (ans != -1) continue; // 已经发现错误，只需读完输入

            if (c[i] == '?') {
                pos.insert(i);
                continue;
            }

            int x = id[i];
            if (a.count(x)) { // 这个券之前出现过
                int last_pos = a[x];
                if (c[last_pos] == c[i]) { // 重复操作（两次I或两次O）
                    // 需要在上次操作之后找一个 '?' 来解释为相反操作
                    auto it = pos.lower_bound(last_pos);
                    if (it == pos.end()) {
                        ans = i; // 无法修复
                    } else {
                        pos.erase(it); // 消耗一个 '?'
                    }
                }
                // 无论是否冲突，更新最近出现位置
                a[x] = i;
            } else { // 这个券第一次出现
                if (c[i] == 'O') { // 第一次出现就是使用，非法
                    // 需要在之前找一个 '?' 解释为 'I x'
                    auto it = pos.lower_bound(0);
                    if (it == pos.end()) {
                        ans = i;
                    } else {
                        pos.erase(it);
                    }
                }
                a[x] = i;
            }
        }

        printf("%d\n", ans);
    }
    return 0;
}
```

## E 任意点

```python
#include <iostream>
#include <cstdio>
#include <algorithm>
using namespace std;

int father[1001];

int find(int x) {
    if (father[x] == x)
        return x;
    else {
        father[x] = find(father[x]);
        return father[x];
    }
}

void merge(int a, int b) {
    int fa = find(a);
    int fb = find(b);
    if (fa != fb) {
        father[fb] = fa;
    }
}

int main() {
    int n;
    while (cin >> n && n) {
        int x[n + 1], y[n + 1], ans = 0;

        // 初始化并查集
        for (int i = 1; i <= n; i++)
            father[i] = i;

        // 读入坐标
        for (int i = 1; i <= n; i++) {
            cin >> x[i] >> y[i];
        }

        // 合并同行或同列的点
        for (int i = 1; i < n; i++) {
            for (int j = i + 1; j <= n; j++) {
                if (x[i] == x[j] || y[i] == y[j]) {
                    merge(i, j);
                }
            }
        }

        // 统计连通分量数量
        for (int i = 1; i <= n; i++) {
            if (father[i] == i)
                ans++;
        }

        // 输出答案
        cout << ans - 1 << endl;
    }
    return 0;
}
```

## F 通配符匹配

```python
#include <iostream>
#include <cstdio>
#include <algorithm>
#include <cstring>
#include <string>
using namespace std;

typedef unsigned long long ull;

const int MAXN = 114514;
ull base[MAXN];           // base[i] = 1331^i
ull ask[MAXN];            // 目标串的前缀哈希
string temp, common;      // temp: 模式串, common: 目标串

struct Wildcard_String {
    ull simple_len[MAXN]; // 每个非通配符段的长度
    ull simple[MAXN];     // 每个非通配符段的哈希值
} wild;

// 预处理模式串：提取非通配符段并计算哈希
void PreTreatMent() {
    int len = 0;
    base[0] = 1;
    for (int i = 1; i < MAXN; ++i)
        base[i] = base[i - 1] * 13331;  // 使用更大的基数减少冲突

    wild.simple[0] = 0;
    for (int i = 0; i < temp.size(); ++i) {
        if (temp[i] == '*' || temp[i] == '?') {
            // 遇到通配符，保存之前累积的段
            wild.simple_len[i - len] = len;
            wild.simple[i - len] = (i > 0 ? wild.simple[i - 1] : 0);
            len = 0;
        } else {
            len++;
            // 更新哈希值
            wild.simple[i] = (i > 0 ? wild.simple[i - 1] * base[1] : 0) + (ull)temp[i];
        }
    }
    // 保存最后一段
    int last_pos = temp.size() - len;
    wild.simple_len[last_pos] = len;
    wild.simple[last_pos] = (temp.size() > 0 ? wild.simple[temp.size() - 1] : 0);
}

// 获取目标串中子串 [l, r] 的哈希值
ull getHash(int l, int r) {
    if (l > r) return 0;
    if (l == 0) return ask[r];
    return ask[r] - ask[l - 1] * base[r - l + 1];
}

// 递归匹配
bool Work(int get_pos, int ask_pos) {
    // 模式串匹配完毕
    if (get_pos >= (int)temp.size()) {
        return ask_pos >= (int)common.size() || temp.back() == '*';
    }
    // 目标串匹配完毕
    if (ask_pos >= (int)common.size()) {
        // 模式串剩余部分必须全是 '*' 或最后一个字符是 '*'
        if (temp.back() == '*' && get_pos == (int)temp.size() - 1) return true;
        for (int i = get_pos; i < (int)temp.size(); ++i) {
            if (temp[i] != '*') return false;
        }
        return true;
    }

    // 处理通配符
    if (temp[get_pos] == '?') {
        return Work(get_pos + 1, ask_pos + 1);
    }
    if (temp[get_pos] == '*') {
        // '*' 匹配0个或多个字符
        for (int i = ask_pos; i <= (int)common.size(); ++i) {
            if (Work(get_pos + 1, i)) return true;
        }
        return false;
    }

    // 处理普通字符段：快速哈希匹配
    int len = wild.simple_len[get_pos];
    if (ask_pos + len > (int)common.size()) return false;

    ull target_hash = getHash(ask_pos, ask_pos + len - 1);
    if (wild.simple[get_pos] == target_hash) {
        return Work(get_pos + len, ask_pos + len);
    }
    return false;
}

int main() {
    ios::sync_with_stdio(false);
    cin.tie(nullptr);

    cin >> temp;
    int t;
    cin >> t;

    PreTreatMent();

    while (t--) {
        cin >> common;
        // 计算目标串前缀哈希
        for (int i = 0; i < (int)common.size(); ++i) {
            ask[i] = (i > 0 ? ask[i - 1] * base[1] : 0) + (ull)common[i];
        }

        if (Work(0, 0)) cout << "YES\n";
        else cout << "NO\n";
    }
    return 0;
}
```

## G 汉诺塔

```python
#include <bits/stdc++.h>
using namespace std;

typedef long long ll;
const int maxn = 31;

int n, g[3][maxn];
ll f[3][maxn];
char s[3];
bool vis[3];

inline void solve() {
    for (int i = 2; i <= n; i++) {
        for (int j = 0; j < 3; j++) {
            int y = g[j][i-1], z = 3 - y - j;
            f[j][i] = f[j][i-1] + 1;
            if (z == g[y][i-1]) {
                f[j][i] += f[y][i-1];
                g[j][i] = z;
            } else {
                f[j][i] += f[y][i-1] + 1 + f[j][i-1];
                g[j][i] = y;
            }
        }
    }
    printf("%lld\n", f[0][n]);
}

inline void init() {
    scanf("%d", &n);
    memset(vis, 0, sizeof(vis));
    for (int i = 0; i < 6; i++) {
        scanf("%s", s);
        int from = s[0] - 'A', to = s[1] - 'A';
        if (vis[from]) continue;
        vis[from] = true;
        g[from][1] = to;
        f[from][1] = 1;
    }
}

int main() {
    init();
    solve();
    return 0;
}
```

## H 马步距离

```python
#include <iostream>
#include <cstring>
#include <algorithm>
using namespace std;

const int dx[8] = {1, 1, -1, -1, 2, 2, -2, -2};
const int dy[8] = {2, -2, 2, -2, 1, -1, 1, -1};

int XP, YP, XS, YS, x, y, ans = 0;
int q[100010][3], head = 0, tail = 1;
int d[110][110];

void init() {
    scanf("%d%d%d%d", &XP, &YP, &XS, &YS);
    x = abs(XP - XS);
    y = abs(YP - YS);
    // 贪心快速逼近
    while (x + y >= 50) {
        if (x < y) swap(x, y);
        if (x - 4 >= 2 * y) x -= 4;
        else x -= 4, y -= 2;
        ans += 2;
    }
    // 偏移坐标，准备BFS
    x += 50;
    y += 50;
}

void BFS() {
    memset(d, -1, sizeof(d));
    q[0][1] = x; q[0][2] = y;
    d[x][y] = 0;
    head = 0; tail = 1;

    while (head != tail) {
        int cx = q[head][1], cy = q[head++][2];
        for (int i = 0; i < 8; i++) {
            int nx = cx + dx[i], ny = cy + dy[i];
            if (nx < 0 || ny < 0 || nx > 100 || ny > 100 || d[nx][ny] != -1) continue;
            d[nx][ny] = d[cx][cy] + 1;
            q[tail][1] = nx; q[tail++][2] = ny;
            if (nx == 50 && ny == 50) return; // 到达起点
        }
    }
}

void work() {
    BFS();
    ans += d[50][50];
    printf("%d\n", ans);
}

int main() {
    init();
    work();
    return 0;
}
```

## I 直方图最大矩形

```python
class Solution {
public:
    /**
     * 代码中的类名、方法名、参数名已经指定，请勿修改，直接返回方法规定的值即可
     *
     * @param heights int整型vector
     * @return int整型
     */
    int largestRectangleArea(vector<int>& heights) {
        int maxm = 0;
        stack<pair<int, int>> s;  // <高度, 左侧不低于该高度的柱子数>
        vector<int> left(heights.size(), 0);  // left[i]记录heights[i]左侧相邻且高度≥它的柱子数

        for (int i = 0; i < heights.size(); i++) {
            if (s.empty() || s.top().first < heights[i]) {
                // 当前高度大于栈顶，直接入栈，左侧延伸数为0
                s.push(make_pair(heights[i], 0));
            } else {
                // 当前高度 <= 栈顶，找到右边界，开始弹出并计算面积
                int right = 1;  // 右侧不低于栈顶高度的柱子数（含自身）
                while (!s.empty() && s.top().first >= heights[i]) {
                    // 累加当前柱子的左侧延伸数
                    left[i] += s.top().second + 1;
                    // 计算以栈顶柱子为高度的最大矩形面积
                    maxm = max(maxm, s.top().first * (right + s.top().second));
                    // 更新right：加上弹出的这个柱子的左侧延伸数和它自身
                    right += s.top().second + 1;
                    s.pop();
                }
                // 将当前柱子压入栈，并记录其左侧延伸数
                s.push(make_pair(heights[i], left[i]));
            }
        }

        // 处理栈中剩余元素（它们的右边界是数组末尾）
        int right = 1;
        while (!s.empty()) {
            maxm = max(maxm, s.top().first * (right + s.top().second));
            right += s.top().second + 1;
            s.pop();
        }

        return maxm;
    }
};
```

## J 消防局的设立

```python
#include <iostream>
#include <vector>
#include <algorithm>
using namespace std;
#define int long long
#define inf 0x3f3f3f3f

const int MAX = 2000;
vector<int> vec[MAX];
int dp[MAX][5]; // dp[u][0..4] 五个状态

void dfs(int u) {
    // 初始化
    dp[u][0] = 1; // 在u建站
    dp[u][3] = 0; // u被子节点覆盖，u自身未覆盖
    dp[u][4] = 0; // u被孙子覆盖，u和子节点可能未覆盖

    for (int son : vec[u]) {
        dfs(son);
        dp[u][0] += dp[son][4]; // u建站，儿子只需保证孙子被覆盖
        dp[u][3] += dp[son][2]; // u被子节点覆盖，儿子需保证自身被覆盖
        dp[u][4] += dp[son][3]; // u被孙子覆盖，儿子需保证其子节点被覆盖
    }

    if (vec[u].empty()) {
        // 叶子节点：没有儿子，只能自己建站来满足状态1和2
        dp[u][1] = dp[u][2] = 1;
    } else {
        dp[u][1] = dp[u][2] = inf;
        // 枚举提供关键覆盖的那个儿子
        for (int i = 0; i < vec[u].size(); i++) {
            int son = vec[u][i];
            // f1: 该儿子建站（状态0），能向上覆盖到u的父节点
            int f1 = dp[son][0];
            // f2: 该儿子的儿子建站（状态1），能向上覆盖到u
            int f2 = dp[son][1];

            for (int j = 0; j < vec[u].size(); j++) {
                if (i == j) continue;
                int other = vec[u][j];
                f1 += dp[other][3]; // 其他儿子只需保证其子节点被覆盖
                f2 += dp[other][2]; // 其他儿子只需保证自身被覆盖
            }
            dp[u][1] = min(dp[u][1], f1);
            dp[u][2] = min(dp[u][2], f2);
        }
    }

    // 状态包含性传递：覆盖强的状态一定满足覆盖弱的状态
    for (int i = 1; i <= 4; i++) {
        dp[u][i] = min(dp[u][i], dp[u][i - 1]);
    }
}

signed main() {
    ios::sync_with_stdio(false);
    cin.tie(nullptr);

    int n;
    cin >> n;
    // 题目输入保证 a[i] < i，所以从2开始建边，1为根
    for (int i = 2; i <= n; i++) {
        int x;
        cin >> x;
        vec[x].push_back(i); // 单向边：父 -> 子
    }

    dfs(1);
    cout << dp[1][2] << endl; // 根节点只需保证自身被覆盖（状态2）

    return 0;
}
```
